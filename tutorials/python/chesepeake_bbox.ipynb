```python
import matplotlib.pyplot as plt
import cartopy.crs as ccrs
import cartopy.feature as cfeature
from matplotlib.patches import Rectangle

# Bounding box
bbox = (-76.75, 36.97, -75.74, 39.01)
lon_min, lat_min, lon_max, lat_max = bbox

# Set up map
fig, ax = plt.subplots(figsize=(8, 6), subplot_kw={"projection": ccrs.PlateCarree()})

# Zoom to bbox region with a little padding
padding = 10.0
ax.set_extent([lon_min - padding, lon_max + padding, lat_min - padding, lat_max + padding], crs=ccrs.PlateCarree())

# Coastlines and borders
#ax.add_feature(cfeature.BORDERS, linestyle=':')
ax.add_feature(cfeature.STATES, linestyle='-', edgecolor='gray')
ax.coastlines(resolution="10m")

# Draw the bounding box
rect = Rectangle(
    (lon_min, lat_min),
    lon_max - lon_min,
    lat_max - lat_min,
    edgecolor="red",
    facecolor="none",
    linewidth=2,
    transform=ccrs.PlateCarree()
)
ax.add_patch(rect)

# Add gridlines with labels
gl = ax.gridlines(draw_labels=True)
gl.top_labels = False
gl.right_labels = False

ax.set_title("Bounding Box Region")
plt.savefig("bbox_chesapeake.png", dpi=300, bbox_inches='tight')
plt.show()
```
